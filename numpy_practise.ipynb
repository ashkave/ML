## using dot product. sales example

```python
sales_array = np.random.randint(200, size=(5,3))
sales_array
```

```python
sales_sheet = pd.DataFrame(sales_array,
                           index =(["Mon","Tue","Wed","thu","Fri"]),
                           columns=(["almond butter", "peanut butter", "cashew butter"]))
                        
```

```python
sales_sheet
```

```python
price_array = np.array([12,8,10])
```

```python
price_array
```

```python
price_sheet = pd.DataFrame(price_array.reshape(1,3),
                           index = ["Price"],
                           columns = ["almond butter", "peanut butter", "cashew butter"])
```

```python
price_sheet
```

```python
total_sales = np.dot(sales_array, price_array)
```

```python
total_sales
```

```python
total_sheet = pd.DataFrame(total_sales.reshape(5,1),
                           index = ["Mon","Tue","Wed","thu","Fri"],
                           columns = ["Total"])
```

```python
total_sheet
```

```python
sales_sheet["Total $"] = total_sheet
```

```python
sales_sheet
```
