## Scikit-Learn Workflow 
1. get the data ready
2. choose a model
3. fit the data into the model (learing patterns)
4. make prediction (using patterns)
5. evaluate the model 
6. experiment to improve accuracy
7. save the trained model for future use 

```python
# 0. standard imports
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

## 1. get the data ready


# get dataset and view dataset
heart_disease = pd.read_csv("heart-disease.csv")
heart_disease.head()

# create X (features) and y (target)

X = heart_disease.drop("target", axis = 1)
y = heart_disease["target"]
X.head(),y.head()

# split the data into train and test set

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y)

X_train.shape, X_test.shape, y_train.shape, y_test.shape

## 2. Choose a model

# choosing an esimator from sk-learn
# random forest classifier is used for classification type of ML problems

from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier()

## 3. Fit the data into the model

model.fit(X_train, y_train)

## 4. Make prediction

y_preds = model.predict(X_test)
y_preds

#single sample. index can change on each run
#X_test.loc[247]
#heart_disease.loc[247]

# Make a prediction on a single sample (has to be array)
#model.predict(np.array(X_test.loc[247]).reshape(1, -1))

## 5. evaulate the model

# on training set
model.score(X_train, y_train)

# on test set (unseen data)

model.score(X_test, y_test)

## 6. check if you can improve

# Try different numbers of estimators (n_estimators is a hyperparameter you can change)
np.random.seed(42)
for i in range(10, 100, 10):
    print(f"Trying model with {i} estimators...")
    model = RandomForestClassifier(n_estimators=i).fit(X_train, y_train)
    print(f"Model accruacy on test set: {model.score(X_test, y_test)}")
    print("")

## 7. save model  and load for future use

import pickle

# Save trained model to file
pickle.dump(model, open("random_forest_model_1.pkl", "wb"))

# Load a saved model and make a prediction on a single example
loaded_model = pickle.load(open("random_forest_model_1.pkl", "rb"))


```
